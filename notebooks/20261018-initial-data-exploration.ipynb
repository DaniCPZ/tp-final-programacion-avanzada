## Conversión de tipos

```python
columnas_porcentuales = [
    '%_PRÉSTAMO_BANCARIO',
    '%_PAGARÉ_BURSÁTIL',
    '%_CHEQUE_de_PAGO_DIFERIDO',
    '%_OTRAS',
    '%_MENOR_a_31_DÍAS',
    '%_MENOR_a_90_DÍAS',
    '%_MENOR_a_180_DÍAS',
    '%_MENOR_a_365_DÍAS',
    '%_MAYOR_a_365_DÍAS'
]

for columna in columnas_porcentuales:
    mora_df[columna] = (
        mora_df[columna]
        .astype(str)
        .str.replace('%', '', regex=False)
        .str.replace(',', '.', regex=False)
        .str.strip()
    )
    mora_df[columna] = pd.to_numeric(mora_df[columna], errors='coerce')
```
